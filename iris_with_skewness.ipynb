```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Iris Dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print("Distribution of", label)
  sns.boxplot(df[label])
  plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

```python
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print("Distribution of", label)
  sns.distplot(df[label],kde=False)
  plt.show()
```

```python
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print("Distribution of", label)
  print(df[label].skew())
```
